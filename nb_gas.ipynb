# Analysis of transfer gas consumption

This notebook compares the gas consumption of ERC-20, ERC-721 and ERC-1155 transfer operations.

```python
colors=['#66c2a5', '#fc8d62', '#648fff']
labels=['ERC-20', 'ERC-721', 'ERC-1155']

erc20_single = 37970
erc721_single = 44282
erc1155_single = 42966

erc1155_batch = [45008, 59273, 71760, 84233, 96707, 159062, 221446, 283803]
num_transfers = [1, 2, 3, 4, 5, 10, 15, 20]

fig = plt.figure(figsize=(9, 3))
ax = fig.add_subplot()
ax.set_title('Transfer gas cost comparison')
ax.set_xlabel('Token transfers')
ax.set_ylabel('Gas cost')
ax.plot(num_transfers, [erc20_single * x for x in num_transfers], color=colors[0], label="ERC-20")
ax.plot(num_transfers, [erc721_single * x for x in num_transfers], color=colors[1], label="ERC-721")
ax.plot(num_transfers, [erc1155_single * x for x in num_transfers], color=colors[2], label="ERC-1155 (single)")
ax.plot(num_transfers, erc1155_batch, color=colors[2], label="ERC-1155 (batch)", linestyle="--")
ax.grid(linestyle='--', linewidth=0.5)
ax.legend()
ax.set_xticks(range(1,21))
utils.set_font_size(ax, 14)
plt.savefig("figures/transfer-gas.pdf", format='pdf', bbox_inches='tight')
plt.show()
```
